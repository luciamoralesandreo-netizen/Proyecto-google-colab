```python
import pandas as pd
#Importo la libreria
```

```python
url = "https://raw.githubusercontent.com/luciamoralesandreo-netizen/Proyecto-google-colab/main/tipos_de_cambio.csv"
#Importo el git con el csv
```

```python
df= pd.read_csv(url)
#Saca los primeros datos del CSV
```

```python
df.head()
```

```python
print("Maximo:", df['Tipo_de_cambio'].max())
print("Minimo:", df['Tipo_de_cambio'].min())
print("Media:", df['Tipo_de_cambio'].mean())
print("Mediana:", df['Tipo_de_cambio'].median())
print("Número de datos:", df['Tipo_de_cambio'].count())
print("Rango:", (df['Tipo_de_cambio'].max()- df['Tipo_de_cambio'].min()))
print("Varianza:", df['Tipo_de_cambio'].var())
print("Desviacion típica:", df['Tipo_de_cambio'].std())
print("Primer cuartil:", df['Tipo_de_cambio'].quantile(0.25))
print("Tercer cuartil:", df['Tipo_de_cambio'].quantile(0.75))
print("Rango intercuantílico:", (df['Tipo_de_cambio'].quantile(0.75) - df['Tipo_de_cambio'].quantile(0.25)))
```

```python
df['Tipo_de_cambio'].describe()
#Otra forma de sacar los datos estadisticos mas facil
```

```python
moneda_max= df.loc[df['Tipo_de_cambio'].idxmax()]
print("Moneda con mayor tipo de cambio:", moneda_max)
moneda_min= df.loc[df['Tipo_de_cambio'].idxmin()]
print("Moneda con menor tipo de cambio:", moneda_min)
```
