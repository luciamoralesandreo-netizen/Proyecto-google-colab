```python
import pandas as pd
import matplotlib.pyplot as plt
#Importo las librerias necesarias
```

```python
url = "https://raw.githubusercontent.com/luciamoralesandreo-netizen/Proyecto-google-colab/main/tipos_de_cambio.csv"
#Importo el git con el csv
```

```python
df= pd.read_csv(url)
```

```python
df
```

```python
monedas=["USD", "GBP", "JPY", "CHF", "CAD", "AUD", "CNY"]
datos= df[df["Moneda"].isin(monedas)]
#Hemos cogido solo algunas monedas porque hay muchas
```

```python
plt.bar(datos["Moneda"], datos["Tipo_de_cambio"])
plt.title("Tipos de cambio")
plt.xlabel("Moneda")
plt.ylabel("Tipo de cambio")
plt.show()
```

```python
plt.hist(df["Tipo_de_cambio"], bins=20)
plt.title("Histograma de tipos de cambio")
plt.xlabel("Tipo de cambio")
plt.ylabel("Frecuencia")
plt.show()
```

```python
plt.boxplot(df["Tipo_de_cambio"])
plt.title("Diagrama de caja de tipos de cambio")
plt.ylabel("Tipo de cambio")
plt.show()
```
